```python
%pip install "nbformat>=4.2.0" ipykernel plotly
import pandas as pd
import plotly.express as px

# 1. Carregamento dos dados
df = pd.read_csv('../Cars dataset.csv')
print("=== PRIMEIRAS LINHAS ===")
display(df.head())

# 2. Inspeção Geral
print("\n=== INFORMAÇÕES DO DATASET ===")
df.info()

print("\n=== VALORES AUSENTES ===")
print(df.isnull().sum())

duplicados = df.duplicated().sum()
print(f"\nLinhas duplicadas encontradas: {duplicados}")

# 3. Tratamento de Duplicatas
# Decisão: Remoção das duplicatas para evitar distorções nas estatísticas
df = df.drop_duplicates()
print(f"Total de registros após remoção de duplicatas: {len(df)}")

print("\n=== RESUMO ESTATÍSTICO ===")
display(df.describe())

# 4. Gráfico 1: Histograma
fig_hist = px.histogram(df, x='selling_price', title='Distribuição dos Preços de Venda')
fig_hist.show()
print("Leitura do Histograma: A distribuição de preços apresenta assimetria à direita, indicando maior concentração de veículos em faixas de valores mais baixas.")

# 5. Gráfico 2: Dispersão (Preço vs Quilometragem)
fig_scatter = px.scatter(df, x='km_driven', y='selling_price', title='Preço vs Quilometragem')
fig_scatter.show()
print("Leitura da Dispersão: Observa-se que veículos com maior quilometragem rodada tendem a apresentar preços de venda menores.")

# 6. Conclusão
print("\nConclusão: A base foi tratada com a remoção de duplicatas e os gráficos confirmam o comportamento esperado de depreciação por quilometragem.")
```

```python
import pandas as pd
import plotly.express as px
import plotly.io as pio

# Configura o renderizador de gráficos para o VS Code
pio.renderers.default = "notebook_connected"

# 1. Carregamento dos dados (caminho relativo)
df = pd.read_csv('../Cars dataset.csv')

print("=== INSPEÇÃO INICIAL ===")
display(df.head())

# 2. Informações, nulos e duplicatas
print("\n=== INFORMAÇÕES DO DATASET ===")
df.info()

print("\n=== VALORES AUSENTES ===")
print(df.isnull().sum())

print("\n=== DUPLICADOS ENCONTRADOS ===")
print(df.duplicated().sum())

# 3. Tratamento de Duplicatas
# Decisão: Remoção de registros duplicados idênticos para evitar distorções
df = df.drop_duplicates()
print(f"\nTotal de linhas após remoção de duplicatas: {len(df)}")

print("\n=== RESUMO ESTATÍSTICO ===")
display(df.describe())

# 4. Histograma
fig_hist = px.histogram(df, x='selling_price', title='Distribuição dos Preços de Venda')
fig_hist.show()
print("Leitura: A distribuição de preços apresenta assimetria à direita, indicando maior concentração em veículos de menor valor.")

# 5. Gráfico de Dispersão
fig_scatter = px.scatter(df, x='km_driven', y='selling_price', title='Preço de Venda vs Quilometragem Rodada')
fig_scatter.show()
print("Leitura: Observa-se que veículos com maior quilometragem rodada tendem a apresentar preços de venda menores.")

# 6. Conclusão
print("\nConclusão: O dataset foi tratado com a remoção de duplicatas e as visualizações confirmam a tendência esperada de depreciação de preço por quilometragem.")
```
